```python
import pandas as pd

# Load the dataset
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Check the number of rows and columns
print("Dataset shape:", df.shape)

# Display the first five rows
display(df.head())

# Check column names and data types
df.info()

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Count customers in each churn category
print(df["Churn"].value_counts())

# Display the percentage of each category
print("\nChurn percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check unique values in each column
for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")
```

```python
# Create a copy of the original dataset
data = df.copy()

# Remove customerID because it is only an identifier
data = data.drop(columns=["customerID"])

# Convert TotalCharges from text to numeric
data["TotalCharges"] = pd.to_numeric(
    data["TotalCharges"],
    errors="coerce"
)

# Check missing values after conversion
print("Missing values after conversion:")
print(data.isnull().sum())

# Check the rows where TotalCharges is missing
print("\nRows with missing TotalCharges:")
display(data[data["TotalCharges"].isnull()])
```

```python
# Separate input features and target variable
X = data.drop(columns=["Churn"])
y = data["Churn"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nNumerical columns:")
print(X.select_dtypes(include=["int64", "float64"]).columns.tolist())

print("\nCategorical columns:")
print(X.select_dtypes(include=["object"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Identify numerical and categorical columns
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

# Numerical preprocessing
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created successfully.")
```

```python
categorical_features = X.select_dtypes(
    include=["str", "object"]
).columns.tolist()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

# Define all five models
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "Naive Bayes": GaussianNB(),

    "Support Vector Machine": SVC(
        probability=True,
        random_state=42
    )
}

# Train each model using the same preprocessing pipeline
trained_models = {}

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    pipeline.fit(X_train, y_train)

    trained_models[name] = pipeline

    print(f"{name} trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

import pandas as pd

results = []

for name, model in trained_models.items():

    # Predict test data
    y_pred = model.predict(X_test)

    # Predict probabilities for AUC
    y_prob = model.predict_proba(X_test)[:, 1]

    # Calculate evaluation metrics
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "AUC": roc_auc_score(
            (y_test == "Yes").astype(int),
            y_prob
        ),
        "Precision": precision_score(
            y_test, y_pred, pos_label="Yes"
        ),
        "Recall": recall_score(
            y_test, y_pred, pos_label="Yes"
        ),
        "F1 Score": f1_score(
            y_test, y_pred, pos_label="Yes"
        ),
        "MCC": matthews_corrcoef(
            y_test, y_pred
        )
    })

# Create comparison table
results_df = pd.DataFrame(results)

display(results_df.round(4))
```

## Model Performance Observations

1. Logistic Regression: Achieved the highest accuracy and AUC, making it the best overall model for this dataset.

2. Decision Tree: Showed the lowest accuracy and AUC among the five models.

3. KNN: Achieved moderate performance, with better recall than Logistic Regression and SVM.

4. Naive Bayes: Achieved the highest recall, making it effective at identifying customers likely to churn. However, its lower precision indicates more false positives.

5. Support Vector Machine: Achieved good accuracy and precision, but its lower recall indicates that it missed more actual churners.

### Overall Winner

Logistic Regression is the best overall model for this dataset because it provides the strongest balance of accuracy, AUC, precision, recall, F1 score, and MCC.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
axes = axes.flatten()

for i, (name, model) in enumerate(trained_models.items()):

    y_pred = model.predict(X_test)

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=["No", "Yes"]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No", "Yes"],
        yticklabels=["No", "Yes"],
        ax=axes[i]
    )

    axes[i].set_title(name)
    axes[i].set_xlabel("Predicted")
    axes[i].set_ylabel("Actual")

fig.delaxes(axes[5])

plt.tight_layout()
plt.show()
```
